```python
%%writefile hello.cu

#include <stdio.h>

__global__ void hello() {
    printf("hello world from gpu! Thread ID : %d\n",threadIdx.x);
}
int main() {
    hello<<<1,5>>>();
    cudaDeviceSynchronize();
    printf("hello world from cpu!\n");
    return 0;
}
```

```python
!nvcc hello.cu -o hello
```

```python
!./hello
```

```python
%%writefile numbers.cu
#include<stdio.h>
__global__ void printNumbers(){
  int i = threadIdx.x+1;
  printf("%d\n",i);
}
int main(){
  printNumbers<<<1,10>>>();
  cudaDeviceSynchronize();
  return 0;
}
```

```python
!nvcc numbers.cu -o printNumbers
```

```python
!./printNumbers
```
