```python
%%writefile cuda3.cu
#include <stdio.h>

__global__ void hello()
{
    printf("Hello from GPU Thread %d\n", threadIdx.x);
}

int main()
{
    hello<<<1,5>>>();

    cudaDeviceSynchronize();

    printf("Hello from CPU\n");

    return 0;
}
```

```python
!nvcc cuda3.cu -o cuda3.cu
```
